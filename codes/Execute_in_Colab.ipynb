```python
import numpy as np
import matplotlib.pyplot as plt

# LOAD TRAINING HISTORY
# Load the loss values saved during training
# These arrays contain the evolution of the error throughout the training epochs
train = np.load("train_loss.npy")
test = np.load("test_loss.npy")

# PLOT LOSS CURVES
# Plot training and testing losses to evaluate the convergence of the model and possible overfitting
plt.figure(figsize=(10, 4))
plt.plot(train, label="Train")
plt.plot(test, label="Test")
plt.xlabel("Epoch")
plt.ylabel("Loss")

# Add grid and legend for better visualization
plt.grid(True)
plt.legend()
plt.show()
```
