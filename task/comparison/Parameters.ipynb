# Configuration File (`default_params.yml`)

This notebook summarizes the modifications made to the original DeepONet configuration file in order to adapt it to the neuronal burst classification problem developed during the internship.

For each modification, the original configuration, the updated version and the motivation behind the change are presented.

## 1. Dataset configuration

The original DeepONet implementation was trained using the Hodgkin–Huxley dataset. It was replaced by the simulated burst datasets generated during this project.

```python
# Original
dataset_train: "dataset/datasetHH_test2_train.mat"
dataset_test: "dataset/datasetHH_test2_test.mat"

# Modified
dataset_train: "dataset/train_bursts.mat"
dataset_test: "dataset/test_bursts.mat"
```

### Changes introduced

- Replaced the original Hodgkin–Huxley datasets.
- Training and testing now use the neuronal burst datasets generated from the four-dimensional ODE model.

## 2. Training configuration

Several hyperparameters were adapted to the size and complexity of the new dataset.

```python
# Original
batch_size: 100
epochs: 10000
scaling: "None"
adapt_actfun: True

# Modified
batch_size: 64
epochs: 300
scaling: "Gaussian"
adapt_actfun: False
optimizer: "Adam"
```

### Changes introduced

- Reduced the batch size to improve GPU memory usage.
- Reduced the number of epochs to shorten training time.
- Added Gaussian normalization.
- Explicitly selected the Adam optimizer.
- Disabled adaptive activation functions.

## 3. DeepONet architecture

The architecture was simplified to better match the new dataset.

```python
# Original
u_dim: 3
G_dim: 700

inner_layer_b: [700,700,700,700]
inner_layer_t: [700,700,700]


# Modified
u_dim: 4
G_dim: 128

inner_layer_b:
  - 256
  - 256

inner_layer_t:
  - 128
  - 128
```

### Changes introduced

- Increased the Branch Network input dimension from three to four variables (`gK0`, `Iapp`, `tau_n` and `eps`).
- Reduced the latent dimension.
- Simplified both Branch and Trunk networks to reduce the number of trainable parameters.

## 4. Logging configuration

```python
# Original
show_every: 5

# Modified
show_every: 10
```

### Changes introduced

Training information is displayed every ten epochs instead of every five epochs, reducing console output while preserving the monitoring of the learning process.

## 5. Parameters kept unchanged

Several parameters from the original implementation were maintained because they were already suitable for the new application:

- DeepONet architecture (`arc = DON`)
- Learning rate (`lr = 0.001`)
- Tanh activation functions
- Layer Normalization (`FNN_LN`)
- Glorot weight initialization
- Trunk input dimension (`x_dim = 1`)
- No Fourier feature mapping (`N_FourierF = 0`)
