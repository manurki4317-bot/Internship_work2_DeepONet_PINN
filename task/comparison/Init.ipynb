## 3. Main modifications

### Removed modules

Several modules from the original repository were removed from the public interface because they are not required for this project.

The removed components include:

- `FNO1d`
- `WNO1d`
- `ResidualBlockCNN`
- `ResNet`
- `MLP`
- `TimeDistributed`
- `myGRU`
- `LinearLayer`
- `AdFNO1d`

These architectures belong to alternative neural network models that are outside the scope of the proposed DeepONet-PINN framework.

### Added modules

Three new functions were incorporated into the package interface.

- **`rhs`**
  - Computes the right-hand side of the Hodgkin–Huxley differential equations.
  - It is used to evaluate the physics residual during PINN training.

- **`m_inf`**
  - Computes the steady-state sodium activation.
  - Used by the Hodgkin–Huxley model.

- **`n_inf`**
  - Computes the steady-state potassium activation.
  - Used by the Hodgkin–Huxley model.

These functions did not exist in the original repository because it was not designed as a Physics-Informed Neural Network.
